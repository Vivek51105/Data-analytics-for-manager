```python
#byjus case
import pandas as pd
import numpy as np
df=pd.read_excel("/content/Lecture2_data.xlsx")
```

```python
# Decision making structure
# Nested IF statement in python
revenue = df["Revenue_Lakh"].mean()
rating= df["Customer_Rating"].mean()

if revenue >=300:
  print("Revenue is high")

  if rating >=4:
    print("Customer satisfaction is also good")
  else:
      print("Customer satisfaction needs improvement")
else:
      print("revenue is low")
```

```python
# Find the average customer rating
rating = df["Customer_Rating"].mean()

# Make a decision based on the rating
if rating >= 4.5:
  print("Excellent")

elif rating >= 4.0:
  print("good")

elif rating >= 3.0:
  print("Average")

else:
  print("needs improvement")
```

```python
# For loop
for rating in df["Customer_Rating"]:

  if rating >= 4:
    print("Good Rating")
  else:
    print("Needs Improvement")
```

```python
# For Loop with range() iloc= integer location

for i in range(10):
  print(df["Product"].iloc[i], df["Revenue_Lakh"].iloc[i])
```

```python
#WHILE Loop in Python
#WHILE loop and i,j is variable to track
i = 0

while i < 5:
  revenue = df["Revenue_Lakh"].iloc[i]

  if revenue >= 300:
    print("High Revenue:", revenue)
  else:
    print("Low Revenue:", revenue)

  i = i + 1
```

```python
import pandas as pd

# Excel file path
FILE = "/content/Starbucks_India_Sales_Dataset.xlsx"

# Load the excel sheets
df = pd.read_excel(FILE, sheet_name="Sales_Data")
products_df= pd.read_excel(FILE,sheet_name="Products")
stores_df= pd.read_excel(FILE,sheet_name="Stores")

# Check the data
print("Sales table:",df.shape)
print("Products:", products_df.shape)
print("Stores:", stores_df.shape)

# Display first 5 rows
df.head()
```

```python
# LIST: the Starbucks menu (8 products)
menu = ["Cappuccino", "Latte", "Americano", "Cold Brew",
        "Frappuccino", "Masala Chai", "Blueberry Muffin", "Veg sandwich"]

# TUPLE: one store's fixed deatails (staoreID ,name, city)
store = ("S01", "Connaught place" ,"Delhi")

#SET : cities where Starbucks sells (the duplicates "Delhi is removed automatically")
cities= {"Delhi","Mumbai","Bengalore","Hyderabad","Pune","Delhi"}

#DICTIONARY: product-> base price in Rs
price= {"Cappuccino":250, "Latte":270,"Americano":220,"Cold Brew":300,
        "Frappuccino":350,"Masala Chai":200,"Blueberry Muffin":180,"Veg Sandwich":240}

print("List     :",menu)
print("Tuple    :",store)
print("Set      :", "<-'Delhi' typed twice,stored once")
print("Dictionary:,price")
print()
print(type(menu),type(store),type(cities),type(price))
```

#Assessing the elements of data structure

```python
# --- List ---
print("First item      :", menu[0])
print("Last item       :", menu[-1])
print("First three     :", menu[0:3])
print("Every 2nd item  :", menu[::2])

# --- Tuple (same as list) ---
print("\nStore ID        :, store[0]")
print("Store city     :",store[2])

#dictionary
print("\nPrice of Latte:",price["Latte"])
print("Safe lookup:",price.get("Pizza","Not on menu"))

#set
print("\nIS Pune a Starbucks city?","Pune" in cities)
print("IS jaipur a Starbucks city?","Jaipur" in cities)
```
